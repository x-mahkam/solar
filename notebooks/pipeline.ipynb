```python
import os
import torch
from torch.utils.data import random_split
from filament_seg.dataset import FilamentDataset

BASE = "/kaggle/input/competitions/filament-segmentation-2026/MAGFiLO_1.0_Kaggle_2026"
full_ds = FilamentDataset(
    f"{BASE}/train/train_images",
    f"{BASE}/train/MAGFiLO_1.0_Annotations_kaggle2026_train.json",
    image_size=512,
)
n_val = max(1, int(len(full_ds) * 0.15))
n_train = len(full_ds) - n_val
generator = torch.Generator().manual_seed(42)
train_subset, val_subset = random_split(full_ds, [n_train, n_val], generator=generator)
val_image_ids = [full_ds.image_ids[i] for i in val_subset.indices]

os.makedirs("/kaggle/working/runs/baseline", exist_ok=True)
with open("/kaggle/working/runs/baseline/val_image_ids.txt", "w") as f:
    f.write("\n".join(str(i) for i in val_image_ids))
val_file_names = [full_ds.images[i]["file_name"] for i in val_image_ids]
with open("/kaggle/working/runs/baseline/val_file_names.txt", "w") as f:
    f.write("\n".join(val_file_names))
print(f"Wrote {len(val_image_ids)} val ids/filenames")
```
